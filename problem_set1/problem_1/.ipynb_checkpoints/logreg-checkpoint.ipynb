```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

class Logsitic_regression():
    def __init__(self, dim, features, objective,
                 OUTPUT_PATH='/home/sssclass/Desktop/project/cs229/problem_set1/output'):
        self.dim = dim
        self.features = features
        self.objective = objective
        self.OUTPUT_PATH = OUTPUT_PATH
        self.e = 10 ** -5

    def sigmoid(self, X, theta):
        return 1 / (1 + np.exp(-X.dot(theta)))

    def predict(self, X, theta, alpha):
        h = self.sigmoid(X, theta)
        return h if alpha is None else h / alpha

    def estimate_alpha(self, validation_data, theta):
        v_pos = validation_data[validation_data[self.objective] == 1]
        return self.sigmoid(v_pos[self.features], theta).mean()
        
    def building_Hessian(self, X, theta):
        h = self.sigmoid(X, theta)
        s = h * (1 - h)
        return -(X.mul(s, axis=0).T).dot(X)

    def calculate_gradient(self, X, y, theta):
        h = self.sigmoid(X, theta)
        return X.T.dot(y - h)

    def newton_optimizer(self, train_data):
        X = train_data[self.features]
        y = train_data[self.objective]
        theta = np.zeros(self.dim)

        while True:
            g = self.calculate_gradient(X, y, theta).values
            H = self.building_Hessian(X, theta).values

            old_theta = theta
            theta = old_theta - np.linalg.solve(H, g)

            if np.abs(theta - old_theta).sum() < self.e:
                break
        return theta

    def ouput_validation(self, validation_data, output_name, theta, alpha=None):
        y_hat = self.predict(validation_data[self.features], theta, alpha)
        y_hat.to_csv(f'{self.OUTPUT_PATH}/{output_name}', index=False, header=False)

    def plot_2d(self, target_plot, plot_name, theta, alpha, ploting = None):
        if ploting is None:
            ploting = self.objective
        fig, ax = plt.subplots(figsize=(12, 5))
        class0 = target_plot[target_plot[ploting] == 0]
        class1 = target_plot[target_plot[ploting] == 1]

        ax.scatter(class0['x_1'], class0['x_2'], marker='o', label='y=0')
        ax.scatter(class1['x_1'], class1['x_2'], marker='x', label='y=1')

        c = 0.0 if alpha is None else np.log(alpha / (2 - alpha))
        xs = np.array([target_plot['x_1'].min(), target_plot['x_1'].max()])
        ys = (c - theta[0] * xs - theta[2]) / theta[1]
        ax.plot(xs, ys, 'r-', label='boundary')

        ax.set_xlabel('x1')
        ax.set_ylabel('x2')
        ax.set_title('logistic prediction')
        ax.legend()
        fig.savefig(f'{self.OUTPUT_PATH}/{plot_name}')
        plt.show()

    def train(self, train_data):
        theta = self.newton_optimizer(train_data)
        return theta

```
